```python
import pandas as pd
import os

BASE = os.path.join(os.getcwd(), "..")  # project root

customers   = pd.read_csv(os.path.join(BASE, "Data", "raw", "customers.csv"))
products = pd.read_csv(os.path.join(BASE, "Data", "raw", "products.csv"))
sellers = pd.read_csv(os.path.join(BASE, "Data", "raw", "sellers.csv"))
orders   = pd.read_csv(os.path.join(BASE, "Data", "raw", "orders.csv"))
order_items = pd.read_csv(os.path.join(BASE, "Data", "raw", "order_items.csv"))
shipment = pd.read_csv(os.path.join(BASE, "Data", "raw", "shipments.csv"))

```

```python
seller_gmv = (order_items.groupby("seller_id")["GMV"]
              .sum()
              .sort_values(ascending=False)
              .reset_index())

seller_gmv["cumulative_pct"] = (seller_gmv["GMV"].cumsum() / 
                                 seller_gmv["GMV"].sum() * 100).round(2)

# How many sellers drive 80% GMV?
top_sellers = seller_gmv[seller_gmv["cumulative_pct"] <= 80]
print(f"Sellers driving 80% GMV : {len(top_sellers)} out of {len(seller_gmv)}")
```
