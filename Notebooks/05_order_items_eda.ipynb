```python
import pandas as pd
import os
```

```python
BASE = os.path.join(os.getcwd(), "..")  # project root
order_items = pd.read_csv(os.path.join(BASE, "Data", "raw", "order_items.csv"))
```

```python
order_items
```

```python
total_order = order_items["order_id"].nunique()

#checking duplicates
dup_count = int(order_items.duplicated(subset="order_item_id").sum())
#checking nulls
missing_val = order_items.isna().sum()
#checking dtype
order_items.dtypes
```

```python
#order_items

print(f"""
========================================
        ORDERS TABLE SUMMARY
========================================
  Total Orders               : {total_order}
  Duplicates                 : {dup_count}
  Missing Values             : {missing_val.any()}
========================================
""")
```

```python
order_items["GMV"]= order_items["quantity"] * order_items["unit_price"]
order_items["platform_revenue"] = order_items["GMV"] * order_items["platform_fee_pct"]
order_items["discount_amount"]  = order_items["GMV"] * order_items["discount_pct"]
order_items["net_GMV"] = order_items["GMV"] - order_items["discount_amount"] 
```

```python
print("Order Items Summary")
print(f"Total GMV            : ₹{order_items['GMV'].sum():,.2f}")
print(f"Total Discount Given : ₹{order_items['discount_amount'].sum():,.2f}")
print(f"Net GMV After Disc   : ₹{order_items['net_GMV'].sum():,.2f}")
print(f"Total Platform Rev   : ₹{order_items['platform_revenue'].sum():,.2f}")

# Multiply by 100 to show as percentage
print(f"Avg Discount %       : {(order_items['discount_pct'].mean()*100).round(2)}%")
print(f"Avg Platform Fee %   : {(order_items['platform_fee_pct'].mean()*100).round(2)}%")
print(f"Avg Order Value      : ₹{order_items['GMV'].mean():,.2f}")
```

```python
to_path = os.path.join(BASE,"DATA","preprocessed")
order_items.to_parquet(f"{to_path}\order_items_clean.parquet")
```
