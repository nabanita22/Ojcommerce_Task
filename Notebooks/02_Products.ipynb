```python
import pandas as pd
import os

BASE = os.path.join(os.getcwd(), "..")  # project root
products = pd.read_csv(os.path.join(BASE, "Data", "raw", "products.csv"))
```

```python
#checking duplicates
dup_count = int(products.duplicated(subset="product_id").sum())
#checking nulls
missing_val = products.isna().sum()
#Number of products
total_prod = products["product_id"].nunique()

#checking the consistency of category and subcategory

uni_cat = list(products["category"].unique())
uni_subcat = list(products["subcategory"].unique())


```

```python
print(f"""
========================================
        PRODUCT TABLE SUMMARY
========================================
  Total Products    : {total_prod}
  Categories        : {len(uni_cat)}
  Sub Categories    : {len(uni_subcat)}
  Duplicates        : {dup_count}
  Missing Values    : {missing_val.any()}
========================================
""")
```

```python
#2. What is the product mix — how many products per category and subcategory?

val = products.groupby(["category","subcategory"])["product_id"].count().reset_index().rename({"product_id":"Prod_count"},axis=1)
val["Total_count_per_cat"] = val.groupby("category")["Prod_count"].transform(sum)
val["pct_subcat"]= round(val["Prod_count"]/val["Total_count_per_cat"],2)*100
cat_dist = val[["category","Total_count_per_cat"]].drop_duplicates()
cat_dist["pct_cat"]= round(cat_dist["Total_count_per_cat"]/cat_dist["Total_count_per_cat"].sum(axis=0),2)*100
cat_dist.sort_values(by="pct_cat",ascending=False)
```

```python
#What is the price distribution across categories and subcategories?
# Are we a budget or premium product marketplace?
# Which category has the highest/lowest base price?
val_price = products.groupby(["category","subcategory"])["base_price"].agg(["min","max","mean","median"]).round(2)
```

```python
#val_price
```

```python
#3. Which categories have the widest price range (premium vs budget products coexisting)?

products.groupby(["category","subcategory"])["base_price"].agg(
    min_price = "min",
    max_price = "max",
    price_range = lambda x : x.max() - x.min(),
    std_dev = "std"
).round(2).sort_values("price_range",ascending=False)
```
